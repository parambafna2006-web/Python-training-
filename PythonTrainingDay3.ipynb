```python
# Writing data to .CSV file

import csv

with open("emp.csv", "w", newline="") as f:
    w = csv.writer(f)

    w.writerow(["ENO", "ENAME", "ESALARY", "EADDR"])

    n = int(input("Enter number of employees : "))

    for i in range(n):
        eno = input("Enter employee no : ")
        ename = input("Enter employee name : ")
        esal = input("Enter employee salary : ")
        eaddr = input("Enter employee address : ")

        w.writerow([eno, ename, esal, eaddr])

print("Data inserted successfully!")

# -----------------------------------------
# Reading data from .CSV file
```

```python
import numpy as np

students = np.array([
    [85, 80, 70],  # Student 1
    [80, 55, 70],  # Student 2
    [85, 92, 88],  # Student 3
    [80, 90, 45],  # Student 4
    [72, 68, 75]   # Student 5
])

students
```

```python
avg_marks = np.mean(students, axis=1)
print("Average Marks :", avg_marks)
```

```python
topper = np.argmax(avg_marks)
print("Topper Students :", topper)
```

```python
topper = np.argmax(avg_marks)
print("Topper :", topper)
```

```python
# Performance Report

for i, avg in enumerate(avg_marks):
    status = "Pass" if avg >= 50 else "Fail"
    print(f"Student {i+1}: Avg = {avg:.2f} | {status}")
```

```python
# Performance Report

print(f"{'Student':<12} {'Average':<12} {'Status':<10}")
print("-" * 34)

for i, avg in enumerate(avg_marks):
    status = "Pass" if avg >= 50 else "Fail"
    print(f"{'Student ' + str(i+1):<12} {avg:<12.2f} {status:<10}")
```

```python
months = ['Jan', 'Feb', 'Mar', 'Apr']
sales = [250,300,400,350]

plt.bar(months, sales)
plt.title("Bar Chart")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.show()
```

```python
plt.barh(months,sales, color='Cyan')
plt.title("Monthly sales(Horizontal)")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.show()
```

```python
#Scatter Plot(Comparing Height vs Weight)
height =[150,160,170,180]
weight =[50,60,70,80]

plt.scatter(height,weight, color='red')
plt.title("Height vs Weight")
plt.xlabel("Height")
plt.ylabel("Weight")
plt.show()
```

```python
#Multiple Plot in one figure
x = [1,2,3,4]
y1 = [2,4,6,8]
y2 = [3,6,9,12]
plt.plot(x,y1,label='Square')
plt.plot(x,y2,label='Linear')
plt.legend()
plt.show()
```

```python
brands = ['A', 'B','C']
shares = [10,20,30]

plt.pie(shares, labels=brands, autopct='%1.1f%%')
plt.title('Market Share')
plt.axis('equal')
plt.show()
```
